```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("modules loaded successfully")
```

```python
df_users = pd.read_csv("../data/users.csv")
df_sessions = pd.read_csv("../data/sessions.csv")
df_events = pd.read_csv("../data/events.csv")
df_orders = pd.read_csv("../data/orders.csv")
df_order_items = pd.read_csv("../data/order_items.csv")

print(f"loaded {len(df_users)} users!")
display(df_users.head())

print(f"loaded {len(df_sessions)} sessions!")
display(df_sessions.head())

print(f"loaded {len(df_events)} events!")
display(df_events.head())

print(f"loaded {len(df_orders)} orders!")
display(df_orders.head())

print(f"loaded {len(df_order_items)} order_items!")
display(df_order_items.head())
```

```python
funnel_counts = df_events['event_name'].value_counts()

print("Raw Event Counts:")
display(funnel_counts)
```

```python
funnel_order = ["search", "view_item_list", "select_item", "view_item", "add_to_wishlist", "share", "add_to_cart", "remove_from_cart", "view_cart", "begin_checkout", "add_shipping_info", "add_payment_info", "purchase", "generate_lead", "refund"] 

ordered_funnel = funnel_counts.reindex(funnel_order)
plt.figure(figsize=(10, 8)) 

sns.barplot(
    x=ordered_funnel.values, 
    y=ordered_funnel.index, 
    hue=ordered_funnel.index, 
    palette="Blues_d", 
    legend=False
)

plt.title("E-commerce Conversion Funnel", fontsize=14, fontweight='bold')
plt.xlabel("Number of Events", fontsize=12)
plt.ylabel("Funnel Stage", fontsize=12)
# plt.xscale('log')
plt.tight_layout()
plt.show()
```

```python
df_merged = df_events.merge(df_sessions[['session_id', 'device_group']], on='session_id', how='left')

device_funnel = df_merged.groupby(['device_group', 'event_name']).size().reset_index(name='count')

plt.figure(figsize=(10, 8))
sns.barplot(
    data=device_funnel, 
    x='count', 
    y='event_name', 
    hue='device_group', 
    order=funnel_order, 
    palette="muted"
)

plt.title("Conversion Funnel: Mobile vs. Desktop", fontsize=14, fontweight='bold')
plt.ylabel("Number of Events")
plt.xlabel("Funnel Stage")
# plt.xscale('log')
plt.show()
```

```python
device_traffic = df_sessions['device_group'].value_counts().reset_index()
device_traffic.columns = ['device_group', 'session_count']

total_sessions = device_traffic['session_count'].sum()
device_traffic['percentage'] = (device_traffic['session_count'] / total_sessions) * 100

plt.figure(figsize=(8, 5))
ax = sns.barplot(
    data=device_traffic, 
    x='device_group', 
    y='session_count',
    hue='device_group', 
    palette=['#4C72B0', '#55A868',] #Blue, Green
)

for i, p in enumerate(ax.patches):
    percentage_text = f"{device_traffic['percentage'].iloc[i]:.1f}%"
    ax.annotate(
        percentage_text, 
        (p.get_x() + p.get_width() / 2, p.get_height()), 
        ha='center', va='bottom', fontsize=12, fontweight='bold', 
        xytext=(0, 5), textcoords='offset points'
    )

plt.title("E-Commerce Traffic Split: Mobile vs. Desktop", fontsize=14, fontweight='bold')
plt.ylabel("Total Shopping Sessions", fontsize=12)
plt.xlabel("Device Platform", fontsize=12)
plt.ylim(0, device_traffic['session_count'].max() * 1.15)
plt.show()
```

```python
df_order_items['total_sales'] = df_order_items['base_item_price'] * df_order_items['quantity']

category_revenue = df_order_items.groupby('item_category')['total_sales'].sum().reset_index()
category_revenue = category_revenue.sort_values(by='total_sales', ascending=False)

plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=category_revenue,
    x='total_sales',
    y='item_category',
    palette='Blues_d',
    hue='item_category'
)

plt.title("Total Sales (Gross Revenue) by Product Category", fontsize=14, fontweight='bold')
plt.xlabel("Total Revenue (₱)", fontsize=12)
plt.ylabel("Product Category", fontsize=12)

plt.xlim(0, category_revenue['total_sales'].max() * 1.2)

total_revenue = category_revenue['total_sales'].sum()

for p in ax.patches:
    width = p.get_width()
    percentage = (width / total_revenue) * 100
    
    label_text = f'₱{width:,.2f} ({percentage:.1f}%)'
    
    plt.text(width + (total_revenue * 0.01), p.get_y() + p.get_height()/2. + 0.05, 
             label_text, ha="left", va="center", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
category_volume = df_order_items.groupby('item_category')['quantity'].sum().reset_index()
category_volume = category_volume.sort_values(by='quantity', ascending=False)

plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=category_volume,
    x='quantity',
    y='item_category',
    palette='Blues_d',
    hue='item_category'
)

plt.title("Total Items Sold by Product Category (Volume)", fontsize=14, fontweight='bold')
plt.xlabel("Total Quantity Sold", fontsize=12)
plt.ylabel("Product Category", fontsize=12)

plt.xlim(0, category_volume['quantity'].max() * 1.4)

total_items = category_volume['quantity'].sum()

for p in ax.patches:
    width = p.get_width()
    percentage = (width / total_items) * 100
    
    label_text = f'{int(width):,} items ({percentage:.1f}%)'
    
    plt.text(width + (total_items * 0.01), p.get_y() + p.get_height()/2. + 0.05, 
             label_text, ha="left", va="center", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
df_sessions['session_start_time'] = pd.to_datetime(df_sessions['session_start_time'])
df_sessions['hour_of_day'] = df_sessions['session_start_time'].dt.hour

hourly_traffic = df_sessions['hour_of_day'].value_counts().reset_index()
hourly_traffic.columns = ['hour_of_day', 'session_count']
hourly_traffic = hourly_traffic.sort_values('hour_of_day')

plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=hourly_traffic, 
    x='hour_of_day', 
    y='session_count',
    hue='hour_of_day',
    palette="magma",
    legend=False
)

plt.title("Shopping Traffic by Hour of Day", fontsize=14, fontweight='bold')
plt.xlabel("Hour of Day (24H Format)", fontsize=12)
plt.ylabel("Total Shopping Sessions", fontsize=12)
plt.ylim(0, hourly_traffic['session_count'].max() * 1.15)

plt.axvline(x=12.5, color='red', linestyle='--', alpha=0.5, label='Lunch Rush')
plt.axvline(x=20.5, color='blue', linestyle='--', alpha=0.5, label='Evening Rush')

plt.legend()

plt.tight_layout()
plt.show()
```

```python
payment_counts = df_orders['payment_type'].value_counts().reset_index()
payment_counts.columns = ['payment_type', 'order_count']

plt.figure(figsize=(10, 5))
ax = sns.barplot(
    data=payment_counts,
    x='order_count',
    y='payment_type',
    palette="crest",
    hue="payment_type"
)

plt.title("Order Volume by Payment Method", fontsize=14, fontweight='bold')
plt.xlabel("Total Orders", fontsize=12)
plt.ylabel("Payment Method", fontsize=12)

plt.xlim(0, payment_counts['order_count'].max() * 1.3)

total_orders = payment_counts['order_count'].sum()

for p in ax.patches:
    width = p.get_width()
    percentage = (width / total_orders) * 100

    label_text = f'{int(width):,} orders ({percentage:.1f}%)'
    
    plt.text(width + (total_orders * 0.01), p.get_y() + p.get_height()/2. + 0.05, 
             label_text, ha="left", va="center", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
df_events_os = df_events.merge(
    df_sessions[['session_id', 'device_operating_system']], 
    on='session_id', 
    how='left'
)

df_android = df_events_os[df_events_os['device_operating_system'] == 'android']

total_android_sessions = df_android['session_id'].nunique()
error_android_sessions = df_android[df_android['android_error'].notna()]['session_id'].nunique()
healthy_android_sessions = total_android_sessions - error_android_sessions

error_rate = (error_android_sessions / total_android_sessions) * 100

plt.figure(figsize=(7, 7))

labels = ['Encountered Error', 'Healthy Sessions']
sizes = [error_android_sessions, healthy_android_sessions]
colors = ['#d62728', '#2ca02c'] # Red, Green

plt.pie(
    sizes, 
    labels=labels, 
    colors=colors, 
    autopct='%1.2f%%', 
    startangle=90,
    pctdistance=0.85, 
    explode=(0.1, 0),
    textprops={'fontsize': 12}
)

centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title(
    f"Android-Specific Checkout Bug Impact\n", 
    fontsize=14, 
    fontweight='bold'
)

plt.tight_layout()
plt.show()

print(f"Total Android Sessions: {total_android_sessions:,}")
print(f"Errored Android Sessions: {error_android_sessions:,}")
print(f"Android Error Rate: {error_rate:.2f}%")
```

```python
purchasing_sessions = df_events[df_events['event_name'] == 'purchase']['session_id'].unique()
df_sessions['made_purchase'] = df_sessions['session_id'].isin(purchasing_sessions)

channel_funnel = df_sessions.groupby('acq_channel').agg(
    total_sessions=('session_id', 'count'),
    total_purchases=('made_purchase', 'sum')
).reset_index()

channel_funnel['conv_rate_pct'] = (channel_funnel['total_purchases'] / channel_funnel['total_sessions']) * 100
channel_funnel = channel_funnel.sort_values('conv_rate_pct', ascending=False)

plt.figure(figsize=(10, 5))
ax1 = sns.barplot(
    data=channel_funnel, 
    x='conv_rate_pct', 
    y='acq_channel', 
    palette="Blues_r",
    hue='acq_channel')

plt.title("Purchase Conversion Rate by Channel", fontsize=14, fontweight='bold')
plt.xlabel("Conversion Rate (%)", fontsize=12)
plt.ylabel("Acquisition Channel", fontsize=12)
plt.xlim(0, channel_funnel['conv_rate_pct'].max() * 1.3)

for p in ax1.patches:
    width = p.get_width()
    idx = int(p.get_y() + p.get_height()/2)
    purchases = int(channel_funnel.iloc[idx]['total_purchases'])
    if width > 0:
        plt.text(width + (width * 0.02), p.get_y() + p.get_height()/2. + 0.05, 
                 f'{width:.2f}% ({purchases:,} orders)', ha="left", va="center", fontsize=11, color='darkblue')

plt.tight_layout()
plt.show()
```

```python
paid_sessions = df_sessions[df_sessions['acq_channel'].isin(['meta', 'tiktok'])]

campaign_funnel = paid_sessions.groupby(['acq_channel', 'utm_campaign']).agg(
    total_sessions=('session_id', 'count'),
    total_purchases=('made_purchase', 'sum')
).reset_index()

campaign_funnel['conv_rate_pct'] = (campaign_funnel['total_purchases'] / campaign_funnel['total_sessions']) * 100
campaign_funnel = campaign_funnel.sort_values(['acq_channel', 'conv_rate_pct'], ascending=[True, False])

plt.figure(figsize=(10, 6))
ax2 = sns.barplot(
    data=campaign_funnel, 
    x='conv_rate_pct', 
    y='utm_campaign', 
    hue='acq_channel', 
    dodge=False, 
    palette="Set2")

plt.title("Conversion Rate by UTM Campaign", fontsize=14, fontweight='bold')
plt.xlabel("Conversion Rate (%)", fontsize=12)
plt.ylabel("UTM Campaign", fontsize=12)
plt.xlim(0, campaign_funnel['conv_rate_pct'].max() * 1.3)

for p in ax2.patches:
    width = p.get_width()
    if width > 0:
        plt.text(width + (width * 0.02), p.get_y() + p.get_height()/2. + 0.05, 
                 f'{width:.2f}%', ha="left", va="center", fontsize=10, color='black')

plt.legend(title='Platform', loc='lower right')
plt.tight_layout()
plt.show()
```

```python
df_order_items['actual_line_revenue'] = df_order_items['quantity'] * df_order_items['final_item_price']
df_order_items['total_discount_given'] = df_order_items['quantity'] * df_order_items['discount_amount']

promo_analysis = df_order_items.groupby('promo_tier_percentage').agg(
    total_transactions=('order_item_id', 'count'),
    avg_quantity=('quantity', 'mean'),
    avg_line_revenue=('actual_line_revenue', 'mean'),
    avg_discount_given=('total_discount_given', 'mean')
).reset_index()

promo_analysis['promo_label'] = (promo_analysis['promo_tier_percentage'] * 100).astype(int).astype(str) + '%'
promo_analysis.loc[promo_analysis['promo_tier_percentage'] == 0.0, 'promo_label'] = 'No Promo (0%)'
promo_analysis = promo_analysis.sort_values('promo_tier_percentage')

fig, ax1 = plt.subplots(figsize=(11, 6))

sns.barplot(
    data=promo_analysis, 
    x='promo_label', 
    y='avg_quantity', 
    palette="Greens",
    hue='promo_label',
    ax=ax1,
    alpha=0.85
)
ax1.set_ylabel('Avg Quantity Purchased (Units)', fontsize=12)
ax1.set_ylim(0, promo_analysis['avg_quantity'].max() * 1.3)

# Add data labels for the bars
for p in ax1.patches:
    ax1.text(p.get_x() + p.get_width()/2., p.get_height() + 0.1, 
             f'{p.get_height():.2f}', ha="center", va="bottom", fontsize=11, fontweight='bold')

ax2 = ax1.twinx()
sns.lineplot(
    data=promo_analysis, 
    x='promo_label', 
    y='avg_line_revenue', 
    color='#2c3e50', # Dark blue
    marker='D', 
    markersize=9, 
    linewidth=3,
    label='Avg Net Revenue (₱)',
    ax=ax2
)

sns.lineplot(
    data=promo_analysis, 
    x='promo_label', 
    y='avg_discount_given', 
    color='#c0392b', # Red
    marker='X', 
    markersize=9, 
    linewidth=3,
    linestyle='--',
    label='Avg Discount Cost (₱)',
    ax=ax2
)

ax2.set_ylabel('Philippine Peso (₱)', fontsize=12, fontweight='bold', color='#2c3e50')
ax2.set_ylim(0, max(promo_analysis['avg_line_revenue'].max(), promo_analysis['avg_discount_given'].max()) * 1.3)

ax2.legend(loc='upper left')

plt.title("Volume Uplift vs. Margin Cannibalization", fontsize=15, fontweight='bold')
ax1.set_xlabel("Voucher Tier Applied", fontsize=12)

plt.tight_layout()
plt.show()

print("\nVOUCHER PERFORMANC")
baseline_rev = promo_analysis[promo_analysis['promo_tier_percentage'] == 0.0]['avg_line_revenue'].values[0]
highest_tier_rev = promo_analysis.iloc[-1]['avg_line_revenue']
highest_tier_label = promo_analysis.iloc[-1]['promo_label']

print(f"Baseline Revenue per line (No Promo): ₱{baseline_rev:,.2f}")
print(f"Revenue per line at {highest_tier_label} discount: ₱{highest_tier_rev:,.2f}")

if highest_tier_rev > baseline_rev:
    print(f"\nCONCLUSION: The {highest_tier_label} tier is PROFITABLE.")
    print("The Poisson-driven increase in quantity purchased successfully offsets the capped peso discount.")
else:
    print(f"\nCONCLUSION: The {highest_tier_label} tier is CANNIBALIZING MARGINS.")
    print("Despite users buying more units, the discount cap is too loose, resulting in a net loss of revenue compared to baseline.")
```

```python
df_order_items['actual_line_revenue'] = df_order_items['quantity'] * df_order_items['final_item_price']
df_order_items['total_discount_given'] = df_order_items['quantity'] * df_order_items['discount_amount']

cohort_df = df_order_items.merge(df_orders[['order_id', 'user_id']], on='order_id', how='left')
cohort_df = cohort_df.merge(df_users[['user_id', 'user_tenure_days']], on='user_id', how='left')

bins = [0, 90, 365, 1000, 2050]
labels = ['Newbies (<90 days)', 'Casuals (90-365 days)', 'Regulars (1-3 years)', 'Veterans (3-5.5 years)']

cohort_df['tenure_cohort'] = pd.cut(cohort_df['user_tenure_days'], bins=bins, labels=labels, right=True)

cohort_metrics = cohort_df.groupby('tenure_cohort', observed=True).agg(
    total_users=('user_id', 'nunique'),
    total_items_bought=('order_item_id', 'count'),
    total_net_revenue=('actual_line_revenue', 'sum'),
    total_discount_cost=('total_discount_given', 'sum')
).reset_index()

cohort_metrics['arpu'] = cohort_metrics['total_net_revenue'] / cohort_metrics['total_users']
cohort_metrics['gross_value'] = cohort_metrics['total_net_revenue'] + cohort_metrics['total_discount_cost']
cohort_metrics['discount_dependency_pct'] = (cohort_metrics['total_discount_cost'] / cohort_metrics['gross_value']) * 100

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

sns.barplot(
    data=cohort_metrics, 
    x='tenure_cohort', 
    y='arpu', 
    color="pink",
    ax=ax1,
    edgecolor='black'
)
ax1.set_title('Average Revenue Per User (ARPU) by Cohort', fontsize=14, fontweight='bold')
ax1.set_ylabel('Net Revenue per User (₱)', fontsize=12)
ax1.set_xlabel('Tenure Cohort', fontsize=12)

sns.barplot(
    data=cohort_metrics, 
    x='tenure_cohort', 
    y='discount_dependency_pct', 
    color="green",
    ax=ax2,
    edgecolor='black'
)
ax2.set_title('Discount Dependency (Voucher Reliance)', fontsize=14, fontweight='bold')
ax2.set_ylabel('% of Gross Value Paid via Discount', fontsize=12)
ax2.set_xlabel('Tenure Cohort', fontsize=12)

plt.tight_layout()
plt.show()

print("\nCOHORT BEHAVIOR")
newbie_dependency = cohort_metrics.loc[cohort_metrics['tenure_cohort'] == 'Newbies (<90 days)', 'discount_dependency_pct'].values[0]
veteran_dependency = cohort_metrics.loc[cohort_metrics['tenure_cohort'] == 'Veterans (3-5.5 years)', 'discount_dependency_pct'].values[0]

print(f"Newbie Discount Dependency: {newbie_dependency:.1f}%")
print(f"Veteran Discount Dependency: {veteran_dependency:.1f}%")

if newbie_dependency > veteran_dependency:
    print("\nINSIGHT: The platform suffers from 'Promo-Hopper' syndrome at the top of the funnel.")
    print("New users heavily rely on vouchers to transact, eating into acquisition margins.")
    print("However, Veterans exhibit strong brand loyalty, buying full-price items and driving the majority of organic platform profitability.")
else:
    print("\nINSIGHT: Retention is purely price-driven.")
    print("Even long-term users refuse to buy without massive discounts. The platform lacks true brand loyalty.")
```
